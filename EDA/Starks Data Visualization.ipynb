### Piechart

```python
data_value = [60, 30, 28, 78, 99]
data_labels = ['A', 'B', 'C', 'D', 'F']
data_colors = ['#3082cf', '#6030cf', '#30cfb5', '#30cf50', '#b430df']

plt.pie(data_value, labels=data_labels, colors=data_colors, wedgeprops={'edgecolor':'black'})
# edge color adds black edges check more in documentation
plt.show()
```

#### Explode Parameter

```python
plt.title('My pie chart :) ')

my_explode = [0.2, 0, 0, 0, 0]    # make all 0 exept data you want to show
plt.pie(data_value, labels=data_labels, colors=data_colors,explode=my_explode , wedgeprops={'edgecolor':'black'})
plt.show()

## ----------------------------------------------------------------- ##

# You can also add shadows

plt.pie(data_value, labels=data_labels, colors=data_colors,explode=my_explode , wedgeprops={'edgecolor':'black'}, 
        shadow=True)
plt.show()

## ----------------------------------------------------------------- ##

# you can change start angles

plt.pie(data_value, labels=data_labels, colors=data_colors,explode=my_explode , wedgeprops={'edgecolor':'black'}, 
        shadow=True, startangle=90)
plt.show()
```

#### Use Hue to visualize the data of different categories in one plot

```python
sns.countplot(x='smoker',hue='marital', data=df2,color='lightblue')
plt.title('Count Plot of Categorical Column')
plt.xticks(rotation=45)
plt.show()
```

#### Lets try to organize the order of the columns

```python
sns.countplot(x='smoker',hue='marital',hue_order=['single','divorced','married'], data=df2,color='lightblue')
plt.title('Count Plot of Categorical Column')
plt.xticks(rotation=45)
plt.show()
```

### Relplot(Realational Plot) used mostly to plot numeric data

https://seaborn.pydata.org/generated/seaborn.relplot.html # link for relplot

```python
#I used heightt to change the length of the plot and aspect to change the width to 1.5* height
sns.relplot(x='duration',y='balance',data=df2,kind='scatter',height=7,aspect=1.5)
plt.show()
```

#### what if you want to make more than 1 plot depending on columns
#### like if col smokers = yes , and col smokers = no 
#### and you want to visualize every col of them with the data

```python
sns.relplot(x='duration',y='balance',data=df2,kind='scatter',col='smoker')
plt.show()
```

#### you can also replace col by row to show different alignment 

```python
sns.relplot(x='duration',y='balance',data=df2,kind='scatter',row='smoker',hue='travelled before')
plt.show()
```

#### and you can use row and col together also


```python
sns.relplot(x='duration',y='balance',data=df2,kind='scatter',col='smoker',row='travelled before')
plt.show()
```

##### now lets asume you want to visualize relation
##### between data x and y through days where day = ["day 1", "day 2", "day 3", "day 4"]

```python
sns.relplot(x="duration",y="balance",data=df2,kind="scatter",col='day',col_wrap=2) ## this will show 2 plots in 1 row
plt.show()
```

```python
sns.relplot(x="duration",y="balance",data=df2,kind="scatter",col='day',col_wrap=4) ## this will show 4 plots in 1 row
plt.show()
```

```python
sns.relplot(x="duration",y="balance",data=df2,kind="scatter",
            col="day",hue='smoker',col_wrap=2,col_order=["day 4","day 3","day 2","day 1"])
plt.show()
## note here that col_wrap parameter will help to make 2 plots in each row while col_order will change the order of the day 
```

##### Style

```python
sns.relplot(x='duration',y='balance',data=df2,kind='scatter',hue='smoker',style='smoker')
plt.show()
## her based on yes, or no there will be 2 colors because of hue
## and based on yes or no for smoker there will be 2 points styles like (x , .)
```

##### transparency

```python
sns.relplot(x='duration',y='balance',data=df2,kind='scatter',alpha=0.5)
plt.show()
## we set alpha parameter between 0 and 1
```

### Catplot (Categorical Plot)

https://seaborn.pydata.org/generated/seaborn.catplot.html #link for catplot

##### count using catplot

```python
sns.catplot(x='Outcome',data=df3,kind='count',palette='magma')
plt.show()
```

```python
sns.catplot(x='Outcome',data=df3,kind='count', order=['1', '0'],palette='magma')
plt.show()
# I changed order
```

#### bar using catplot

```python
sns.catplot(x='BloodPressure',y='Insulin',data=df3,kind='bar',palette='rocket')
sns.catplot(x='Outcome',y='Insulin',data=df3,kind='bar',palette='magma')
plt.show()
```

#### box using catplot plot

```python
sns.catplot(x='BloodPressure',y='Insulin',data=df3,kind='box',palette='rocket')
sns.catplot(x='Outcome',y='Insulin',data=df3,kind='box',palette='Blues')
plt.show()
```

#### Line using catplot

```python
sns.catplot(x='Outcome',y='Insulin',data=df3,kind='point')
# if you want to remove lines connecting the points make join = False
sns.catplot(x='Outcome',y='Pregnancies',data=df3,kind='point',color='red',join=False)
# one of the good things that you can even make ci using median not mean and you can ci = None
sns.catplot(x='Outcome',y='Glucose',data=df3,kind='point',color='green',ci=None)
plt.show()
```

```python
sns.set_style("whitegrid")
sns.relplot(x='Pregnancies',y='Glucose',data=df3,kind='line', style='Outcome', hue='Outcome', dashes=False, ci=None)
plt.show()
```
