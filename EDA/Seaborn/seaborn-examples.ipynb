```python
from matplotlib import pyplot as plt
import seaborn as sns
import warnings
import numpy as np
warnings.simplefilter('ignore')
```

```python
cars=sns.load_dataset('mpg').dropna()
```

```python
cars
```

# KDE
kernel density estimation


it allows us to estimate the probability density function from the data 

to build it imagine that you have number of points in  each pont add little centerd kernel at the end add all of them together it will make the kernel density estimation

## Univariate(1D)
just for one variable

```python
sns.kdeplot(cars.horsepower)
```

```python
#shade
#if you  need to shade underneath the KDE you can turn  shade on
sns.kdeplot(cars.horsepower,shade=True)
```

```python
#bandwidth
#if you decrease it you will see more jagged peaks and it increases the variance of KDE
#if you increase it you will see curves smooth out and it will lose the mode in data 
sns.kdeplot(cars.horsepower,bw=20)
```

```python
# cumulative
# if you want to switch to  cumulative distribution function instead of probability turn it on 
sns.kdeplot(cars.horsepower,cumulative=True)
```

## Bivariate (2D)
for two variables it allows you to create contors representing various different density levels of your data 

```python
sns.kdeplot(
    data=cars, x='horsepower', y='mpg',
)
```

```python
# n_levels
# if you have more rings 
sns.kdeplot(
    data=cars, x='horsepower', y='mpg',n_levels=20
)
```

```python
# shade
# it turn from lines to a shaded version 
sns.kdeplot(
    data=cars, x='horsepower', y='mpg',shade=True
)
```

```python
# cbar
# if you want to see the color bar
sns.kdeplot(
    data=cars, x='horsepower', y='mpg',shade=True,cbar=True
)
```

# Histogarm

```python
sns.histplot(data=cars,x='horsepower')
```

```python
# if you want to make horizontal bars instead of vertical put it in y
sns.histplot(data=cars,y='horsepower')
```

```python
# kde
# if you want to put kernel density estimation on the histogram
sns.histplot(data=cars,x='horsepower',kde=True)
```

```python
# bins
# if you want to put your bins instead of histo pins
sns.histplot(data=cars,x='horsepower',bins=20)
```

```python
# binwidth and binrange
# binwidth its for the width of the bins 
# binrange its for the range for start and end
sns.histplot(data=cars,x='horsepower',binwidth=1,binrange=(5,10))
```

```python
# stat
# its for statistics the default for the histogram is count it apper on y axis we can change it to:
# density: area of histogram sum to one
# probability: height of histogram bars to one 
sns.histplot(data=cars,x='horsepower',stat='density')
```

```python
# hue
# if you have categorical variables and want to visulize it on the data 
sns.histplot(data=cars,x='horsepower',hue='origin')
```

```python
# element 
# its an option you can change to easy read the results you can put it to:
# step: it make easier to distinguish the diffrent layers of data
# poly: it will make bars more smooth
sns.histplot(data=cars,x='horsepower',hue='origin',element='step')
```

```python
# multiple
# it show how the bars look like you have:
# stack: it stack the bars on each other
# fill: it scale each bucket to range between zero and one and it show show the proportion of each specie
sns.histplot(data=cars,x='horsepower',hue='origin',multiple='stack')
```

# Heatmap

```python
origin_cyl=cars.groupby('origin').cylinders.value_counts().unstack().fillna(0)
```

```python
sns.heatmap(origin_cyl)
```

```python
# trasnpose
# it will trasnpose your heatmap
sns.heatmap(origin_cyl.T)
```

```python
# correlation
# it show the correlation  between values in the heatmap
sns.heatmap(origin_cyl.corr())
```

```python
# cmap
# to change the color in your heatmap
sns.heatmap(origin_cyl,cmap="Blues")
```

```python
# annotations
# to add annotations to heatmap 
# fmt
# to decide number of decimal points in the number
sns.heatmap(origin_cyl,cmap="Blues",annot=True,fmt='.0f')
```

```python
# linewidth
# to add lines between boxes
sns.heatmap(origin_cyl,cmap="Blues",annot=True,fmt='.0f',linewidth=1)
```

# countplot vs barplot
countplot: we just count the number of observations per category
barplot: we see summary statistcs on the bars like avg or sum like that

```python
# Order ascending or descending
# if you want to order your values
#descending
sns.countplot(data=cars,x='origin',order=cars.origin.value_counts().index)
```

```python
# ascending
sns.countplot(data=cars,x='origin',order=cars.origin.value_counts().index[::-1])
```

```python
#hue
# if you want to handle another type of  data
sns.countplot(data=cars,x='origin',hue='cylinders')
```
