```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("adult.csv")
data.head()
```

```python
data.describe(include='object')
```

```python
data['income'].value_counts()
```

```python
categorical_data = []
numerical_data = []

for col in data:
    if data[col].dtype == "object":
        categorical_data.append(col)
    else:
        numerical_data.append(col)

categorical_data.remove('income')

print(categorical_data)
print(numerical_data)
```

```python
from sklearn.model_selection import train_test_split

x = data.drop('income',axis=1)
y = data[['income']]

X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder, StandardScaler

le = LabelEncoder()
enc = OrdinalEncoder()
sc = StandardScaler()
```

```python
X_train[categorical_data] = enc.fit_transform(X_train[categorical_data])
X_test[categorical_data] = enc.transform(X_test[categorical_data])

X_train[numerical_data] = sc.fit_transform(X_train[numerical_data])
X_test[numerical_data] = sc.transform(X_test[numerical_data])

y_train['income'] = le.fit_transform(y_train['income'])
y_test['income'] = le.transform(y_test['income'])
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score
from sklearn.dummy import DummyClassifier

dummy_clf = DummyClassifier(strategy='most_frequent')
dummy_clf.fit(X_train,y_train)
preds = dummy_clf.predict(X_test)
print(f"dummy classifier accuracy is : {accuracy_score(preds,y_test)}")
print(f"dummy classifier f1_score is : {f1_score(preds,y_test)}")
```

```python
rf = RandomForestClassifier()
rf.fit(X_train,y_train)
preds = rf.predict(X_test)
print(f"random forest accuracy is : {accuracy_score(preds,y_test)}")
print(f"random forest f1_score is : {f1_score(preds,y_test)}")
```
